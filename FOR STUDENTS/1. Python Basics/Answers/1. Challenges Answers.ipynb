# Rock, Paper, Scissors!

In this challenge, you will make a game of rock paper scissors against a bot!
- The player inputs either `"rock"`, `"paper"`, or `"scissors"` (or 0, 1, 2 to represent rock, paper, scissors respectively).
- The bot will randomly choose their hand.
- Print the bot's hand and display whether it is a draw or a win by the player or bot  
*Hint: `random` is a library with a host of functions related to random chance. It will be helpful here!*

````python
import random

player_hand = input("Enter your hand (rock/paper/scissors): ").strip().lower()

bot_number = random.randint(0, 2) # 0=rock, 1=paper, 2=scissors

if bot_number == 0:
    bot_hand = "rock"
elif bot_number == 1:
    bot_hand = "paper"
else:
    bot_hand = "scissors"

print("Bot chose: " + bot_hand)

if player_hand == bot_hand:
    print("It's a draw!")
elif player_hand == "rock" and bot_hand == "scissors":
    print("You win!")
elif player_hand == "paper" and bot_hand == "rock":
    print("You win!")
elif player_hand == "scissors" and bot_hand == "paper":
    print("You win!")
else:
    print("Bot wins!")
````

````python
# You can actually do most of this in one line (interesting but not very practical!)
import random

player_number = int(input("Enter your hand using a number (0=rock/1=paper/2=scissors): ").strip())

bot_number = random.randint(0, 2) # 0=rock, 1=paper, 2=scissors

if bot_number == 0:
    bot_hand = "rock"
elif bot_number == 1:
    bot_hand = "paper"
else:
    bot_hand = "scissors"

print("Bot chose: " + bot_hand)

print(["It's a draw!", 'You win!', 'Bot wins!'][(player_number - bot_number) % 3]) # Uses the concept of lists
````

## 🛠️ The Mission Requirements

Write a script that does the following **in order**:

1. **Welcome the player**: Ask for their name using `input()`. Print a welcome message.

2. **The Shop**: Tell the player they have arrived at a shop. Display their current gold. A sword costs `30` gold and a shield is **free**, with a catch 🤔.
   - Ask: `"Do you want to buy a Sword? (yes/no) "`
     - If yes AND they have enough gold: subtract the cost, set `has_sword` to `True`, and print a success message.
     - If yes but not enough gold: print `"Not enough gold!"`.
   - Ask: `"Do you want a Shield? (yes/no) "`
     - If yes, the merchant wants to test the player's skills at rock paper scissors!
     - Copy your code over from the above challenge and modify it to fit the scenario.
     - If the player wins, they can take the shield. Update `has_shield`.  
     - If the merchant wins, the player cannot take the shield.
     - If it is a draw, the merchant randomly decides whether to give you the shield for your valiant efforts!
   - Print their updated `gold`, `has_sword`, and `has_shield`.

3. **The Crossroads**: The player reaches a fork in the road. Ask them to choose `"left"` or `"right"`.
   - **Left path**: The player finds a healing spring! Add `25` to their health (but health should not go above `100`). Print the result.
   - **Right path**: The player encounters a bandit!
     - If `has_sword` is `True`: print `"You fight the bandit and win 20 gold!"` and add `20` gold.
     - Else: print `"The bandit steals 15 gold!"` and subtract `15` gold.
   - If neither `"left"` nor `"right"`: print `"You stand still, confused."`

4. **Final Stats**: Print the player's final `health`, `gold`, `has_sword`, and `has_shield`.

````python
# Starting stats
gold = 50
health = 85
has_sword = False
has_shield = False

import random

name = input("Enter your name: ").strip()
print(f"Welcome, {name}! Your adventure begins...")

print("\n--- You arrive at a shop! ---")
print(f"Gold: {gold}")

# Sword
sword_choice = input("Do you want to buy a Sword? (yes/no) ").strip().lower()
if sword_choice == "yes":
    if gold >= 30:
        gold -= 30
        has_sword = True
        print("You bought a sword!")
    else:
        print("Not enough gold!")

print(f"\nGold: {gold} | Sword: {has_sword} | Shield: {has_shield}")

# Shield (via Rock Paper Scissors)
shield_choice = input("\nDo you want a Shield? (yes/no) ").strip().lower()
if shield_choice == "yes":
    print("The merchant challenges you to Rock Paper Scissors for the shield!")

    player_hand = input("Enter your hand (rock/paper/scissors): ").strip().lower()

    bot_number = random.randint(0, 2) # 0=rock, 1=paper, 2=scissors

    if bot_number == 0:
        bot_hand = "rock"
    elif bot_number == 1:
        bot_hand = "paper"
    else:
        bot_hand = "scissors"

    print("Merchant chose: " + bot_hand)

    if player_hand == bot_hand:
        print("It's a draw! The merchant flips a coin...")
        has_shield = random.randint(0, 1) == 1
        if has_shield:
            print("The merchant gives you the shield for your valiant efforts!")
        else:
            print("The merchant keeps the shield because you are not worthy.")

    elif player_hand == "rock" and bot_hand == "scissors":
        print("You win! You take the shield.")
        has_shield = True
    elif player_hand == "paper" and bot_hand == "rock":
        print("You win! You take the shield.")
        has_shield = True
    elif player_hand == "scissors" and bot_hand == "paper":
        print("You win! You take the shield.")
        has_shield = True
    else:
        print("Merchant wins! No shield for you.")

print(f"\nGold: {gold} | Sword: {has_sword} | Shield: {has_shield}")

# The Crossroads
print("\n--- You reach a crossroads! ---")
path = input("Do you go left or right? ").strip().lower()

if path == "left":
    health = min(100, health + 25)
    print(f"You find a healing spring! Health is now {health}.")
elif path == "right":
    print("A bandit appears!")
    if has_sword:
        print("You fight the bandit and win 20 gold!")
        gold += 20
    else:
        print("The bandit steals 15 gold!")
        gold -= 15
else:
    print("You stand still, confused.")

# Final Stats
print("\n--- Final Stats ---")
print(f"Health: {health}")
print(f"Gold:   {gold}")
print(f"Sword:  {has_sword}")
print(f"Shield: {has_shield}")
````
