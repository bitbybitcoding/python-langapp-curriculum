# **File IO Tutorial!**

## An important function for dealing with files in Python is open().

### There are 4 primary ways to interact with files:
- "w", which is for writing/overwriting
- "r", which is for reading
- "a", which is for appending content
- "x", which is for creating files

### Moreover, the 2 types of files you may deal with are
- "t", which are text files
- "b", which are binary files such as images

# Section 1 : Opening Files
This is how one can use open(). The file value will be returned so ensure you store it in a variable as shown below.

Write mode creates the file if it does not already exist, and OVERWRITES the file if it does already exist.

Therefore, be extremely cautious when using it instead of append or create.

Remember to close your file after opening it and using it. This frees memory and prevents corruption.

```python
file = open("filename.txt", "w") # You may replace "w" with "r", "a" or "x"
file.write('hi')
file.close()
```

# Section 2 : With Open & Basic Reading
This automatically closes the file after the actions have been completed, saving you time. If you are taking H2 Computing, you MUST specify you understand this and comment the fact that the file will be automatically closed. If not, the marks will NOT be awarded to you.

This program uses read(), which allows the contents of the file to be stored into the variable as shown below.

```python
with open("filename.txt", "r") as f: # open() automatically closes the file
    content = f.read()
```

# Section 3 : Advanced Reading
Below shows 3 alternative ways to read, allowing you to specify what part of the file you want read, and how the data returned should be formatted. This can help a lot in data cleaning.

```python
with open("filename.txt", "r") as f:
    f.read(5) # returns certain number of characters from the file
    f.readline() # returns one line at a time
    lines = f.readlines() # returns a list of all lines
```

# Section 4 : Advanced Writing
Below shows the 2 main ways to write to a file.

```python
with open("data.txt", "w", encoding="utf-8") as f:
    f.write("Hello World") # this overwrites the entire file, but write() only accepts strings
```

```python
lines = ["Apple\n", "Banana\n", "Cherry\n"]
with open("data.txt", "w") as f:
    f.writelines(lines) # the input must include \n when using writelines to write multiple lines
```

# Section 5 : Appending
Appending is similar to writing, as it is an input method. However, as it only adds to the end of a file, it is considered safer to use.

This still uses write(), but the type in open() is "a".

```python
with open("data.txt", "a") as f:
    f.write("New line\n") # note how write() is still used, but the open() uses "a" and not "w"
```

# Section 6 : Data Cleaning
To clean data, you will need to use .strip() and then .split().

.strip() and other variants deal with the spaces and endlines that can make the string unclean, and uniformly removes all of them.

.split() splits up the string into a list of smaller strings based on a specified character. Usually, this is " ", a blank space. However, if you are dealing with CSV files, you can use ",".

Sometimes the file types given are weird, and splitting the strings isn't very easy. When this is the case, keep experimenting to find the best character by which you can split the string, and then use it.

```python
with open("users.txt") as f:
    for line in f:
        clean = line.strip()
        parts = clean.split(",")
        print(parts)
```
