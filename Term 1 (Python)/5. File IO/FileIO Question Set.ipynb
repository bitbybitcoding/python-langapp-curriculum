# File IO Questions

Run the cell below to generate the files you need for your work. Refer to the section headers to see what questions are to be done in what setting. Have fun!

```python
import random
import csv

random.seed(42)

names = [
    "Alice","Ben","Chloe","Daniel","Ethan","Farah","Grace","Hassan","Ivy","Jason",
    "Kiran","Liam","Maya","Noah","Olivia","Priya","Quinn","Ryan","Sara","Tom"
]

with open("student_scores.txt", "w") as f:
    for name in names:
        score = random.randint(40, 100)
        f.write(f"{name} {score}\n")

with open("attendance_log.txt", "w") as f:
    for name in names:
        days = random.randint(10, 20)
        f.write(f"{name},{days}\n")
messages = [
    "Python makes file handling simple.",
    "Students must learn how to read and write files.",
    "Data analysis often starts with loading files.",
    "Errors can occur when files are missing.",
    "Practice improves programming skills.",
    "Always close files or use the with statement.",
    "CSV files are widely used in data science.",
    "Debugging file paths can be frustrating.",
]

with open("messages.txt", "w") as f:
    for line in messages:
        f.write(line + "\n")

items = ["Pen","Notebook","Calculator","Ruler","Marker","Folder","Stapler"]

with open("sales_data.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["Item","Price","Quantity"])

    for item in items:
        price = round(random.uniform(1.5, 15.0), 2)
        quantity = random.randint(5, 50)
        writer.writerow([item, price, quantity])

print("Files created: student_scores.txt, attendance_log.txt, messages.txt, sales_data.csv")
```
