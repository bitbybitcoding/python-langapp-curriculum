# Object-Oriented Programming (OOP)

Welcome to one of the biggest milestones in coding! Up until now, we’ve written lists of instructions. With **OOP**, we build virtual "things" that have their own data and behaviors.

Think of it like a video game: instead of just writing code for "health" and "speed," you create a **Player** object that carries its own health, speed, and weapons wherever it goes.

## 🏗️ Classes vs. Objects

- **Class (The Blueprint)**: This is the design. It says all players should have a name and a level, but it isn't a specific person yet.
- **Object (The Instance)**: This is the actual character you play as. If the class is the "Cookie Cutter," the object is the "Cookie."

### 🏷️ Attributes and Verbs
- **Attributes (Adjectives)**: Variables that belong to the object (e.g., `hp = 100`).
- **Methods (Verbs)**: Functions that belong to the object (e.g., `.jump()` or `.attack()`).

## 🧬 Inheritance: Leveling Up

What if you want a special type of classroom, like a **ProjectGroup**?

Instead of rewriting all the code for teachers and students, we can **Inherit** from our original class. This makes the `ProjectGroup` a "Child" and the `BitByBitClass` the "Parent."

- Use `super().__init__()` to grab everything from the parent's blueprint.

*Pro Tip: The parent can't access any attributes or methods from the child!*

```python
class ProjectGroup(BitByBitClass):
    def __init__(self, teacher_name, project_name):
        # Inherit the teacher and student logic from the parent
        super().__init__(teacher_name)
        self.project = project_name
        print(f"Project group assigned to: {self.project}")

# Create a project group
coding_squad = ProjectGroup("Mrs. Tan", "Build a Game")
```

## 🔍 The OOP Summary

| Concept | Analogy | Python Example |
| :--- | :--- | :--- |
| **Class** | The Blueprint / Recipe | `class Dog:` |
| **Object** | The actual House / Cookie | `my_dog = Dog()` |
| **Attribute** | The Specs (Color, Size) | `self.health = 100` |
| **Method** | The Actions (Run, Bark) | `def attack(self):` |
| **Inheritance** | A specialized version | `class RobotDog(Dog):` |
